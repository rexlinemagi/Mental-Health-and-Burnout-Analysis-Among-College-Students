```python
import pandas as pd
import numpy as np

df = pd.read_csv("Mental Health and Burnout Analysis Among College Students.csv")
print("Original Shape:", df.shape)
```

HANDLE NULL VALUES

```python
print("Handling Null Values")
if "Age" in df.columns and df["Age"].isnull().sum() > 0:
    avg_age = int(df["Age"].mean())
    df['Age'] = df['Age'].fillna(avg_age)
    print("Filled 'Age' nulls with the mean value:", avg_age)
else:
    print("'Age' column has no null values to fill.")
```

CASE STANDARDIZATION

```python
print("Case Standardization")
text_columns = [
    "Name (or Nickname)", "Discipline/Stream of Education", "Year of Study"
]
for col in text_columns:
    if col in df.columns:
        df[col] = df[col].astype(str).str.lower().str.strip()
print("Standardized case for text columns.")
```

NON-NUMERIC TO NUMERIC CONVERSION

```python
print("Non-Numeric to Numeric Conversion")

# Mapping for comfort talking about mental health
comfort_map = {"no, never": 0, "rarely": 1, "sometimes": 2, "yes, always": 3}
df['Do you feel comfortable talking to friends/family about your mental health struggles?'] = df['Do you feel comfortable talking to friends/family about your mental health struggles?'].astype(str).str.lower().str.strip().map(comfort_map)

# Mapping for year of study
year_map = {'1st year': 1, '2nd year': 2, '3rd year': 3, '4th year': 4}
df['Year of Study'] = df['Year of Study'].map(year_map)

# Mapping for academic hours
hours_map = {'less than 10 hours': 5, '10–20 hours': 15, '20–30 hours': 25, 'more than 30 hours': 35}
df['How many hours per week do you spend on academics (classes + studying)?'] = df['How many hours per week do you spend on academics (classes + studying)?'].astype(str).str.lower().str.strip().map(hours_map)

# Mapping for sleep hours
sleep_map = {'less than 5 hours': 4, '5–6 hours': 5.5, '6–7 hours': 6.5, '7–8 hours': 7.5, 'more than 8 hours': 9}
df['On average, how many hours of sleep do you get per night?'] = df['On average, how many hours of sleep do you get per night?'].astype(str).str.lower().str.strip().map(sleep_map)

# Mapping for skipping meals
freq_map = {'never': 0, 'rarely': 1, 'sometimes': 2, 'often': 3, 'always': 4}
df['How often do you skip meals or eat irregularly because of workload or stress?'] = df['How often do you skip meals or eat irregularly because of workload or stress?'].astype(str).str.lower().str.strip().map(freq_map)

# Mapping for social media use
df['How often do you use social media or digital platforms as a way to cope with stress?'] = df['How often do you use social media or digital platforms as a way to cope with stress?'].astype(str).str.lower().str.strip().map(freq_map)

# Mapping for exercise frequency
exercise_map = {'never': 0, 'rarely': 1, 'occasionally': 2, '2-3 times a week': 3, 'daily': 4}
df['Do you engage in any form of exercise or physical activity?'] = df['Do you engage in any form of exercise or physical activity?'].astype(str).str.lower().str.strip().map(exercise_map)

# Mapping for yes/no/maybe style questions
pressure_map = {'no': 0, 'maybe/not sure': 1, 'yes': 2}
df['Has academic pressure ever affected your mental health severely (e.g., anxiety, depression, breakdowns)?'] = df['Has academic pressure ever affected your mental health severely (e.g., anxiety, depression, breakdowns)?'].astype(str).str.lower().str.strip().map(pressure_map)
support_map = {'no': 0, 'somewhat': 1, 'yes': 2}
df['Do you think your college provides enough support/resources for student mental health?'] = df['Do you think your college provides enough support/resources for student mental health?'].astype(str).str.lower().str.strip().map(support_map)

print("Mapping complete.")
```

```python
# Label Encode 'Discipline/Stream of Education'
disciplines = df['Discipline/Stream of Education'].unique()
discipline_map = {discipline: i for i, discipline in enumerate(disciplines)}
df['Discipline/Stream of Education'] = df['Discipline/Stream of Education'].map(discipline_map)
print("'Discipline' column label encoded.")
```

```python
# Extract the first coping mechanism only
coping_col = 'What coping mechanisms do you use when feeling stressed or burned out?'
df['main_coping_mechanism'] = df[coping_col].astype(str).str.split(';').str[0].str.strip()

# Label Encode the new simplified column
coping_mechanisms = df['main_coping_mechanism'].unique()
coping_map = {mechanism: i for i, mechanism in enumerate(coping_mechanisms)}
df['main_coping_mechanism'] = df['main_coping_mechanism'].map(coping_map)
print("'Coping Mechanisms' column simplified and label encoded.")
print("Coping Map:", coping_map)

# Drop the original text column for coping mechanisms
df.drop(columns=[coping_col], inplace=True)
```

FINAL CLEANUP

```python
print("Final Cleanup")
columns_to_drop = ['Unnamed: 0', 'Timestamp', 'Name (or Nickname)']
df.drop(columns=columns_to_drop, inplace=True, errors='ignore')
print("Dropped final unnecessary columns:", columns_to_drop)
```

SAVE PROCESSED DATA

```python
print("Saving Process Data")
df.to_csv("mental_health_processed.csv", index=False)
print("SUCCESS: Processed data saved to 'mental_health_fully_processed.csv'")

# Display final DataFrame information
print("Final Data Info:\n")
df.info()
```
