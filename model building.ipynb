Random Forest Classifier

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Load dataset
df = pd.read_csv('mental_health_processed.csv')

target_column = 'On a scale of 1–5, how would you rate your overall mental well-being right now?'

# Create new binary target column
# If the rating is 3 or less, it's 0 (Not Good). If it's more than 3 (i.e, 4 or 5), it's 1 (Good).
df['well_being_binary'] = np.where(df[target_column] <= 3, 0, 1)

# Set features (df_x) and NEW target (df_y)
df_x = df.drop(columns=[target_column, 'well_being_binary']) # Drop old and new target
df_y = df['well_being_binary'] # New target

# Split data
df_x_train, df_x_test, df_y_train, df_y_test = train_test_split(df_x, df_y, test_size=0.2, random_state=50)

# Initialize and train the model
rf = RandomForestClassifier(random_state=42)
rf.fit(df_x_train, df_y_train)

# Predict and get the accuracy
pred_rf = rf.predict(df_x_test)
accuracy = accuracy_score(df_y_test, pred_rf) * 100

print("Random Forest Accuracy:")
print(accuracy)
```

K-Nearest Neighbour

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Load dataset
df = pd.read_csv('mental_health_processed.csv')

target_column = 'On a scale of 1–5, how would you rate your overall mental well-being right now?'

# Create new binary target column
# If the rating is 3 or less, it's 0 (Not Good). If it's more than 3 (i.e, 4 or 5), it's 1 (Good).
df['well_being_binary'] = np.where(df[target_column] <= 3, 0, 1)

# Set features (df_x) and NEW target (df_y)
df_x = df.drop(columns=[target_column, 'well_being_binary']) # Drop old and new target
df_y = df['well_being_binary'] # New target

# Split data
df_x_train, df_x_test, df_y_train, df_y_test = train_test_split(df_x, df_y, test_size=0.2, random_state=50)

# Initialize KNN model
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(df_x_train, df_y_train)

# Predict and get the accuracy
pred_knn = knn.predict(df_x_test)
accuracy = accuracy_score(df_y_test, pred_knn) * 100

print("KNN Accuracy:")
print(accuracy)
```

Decision Tree

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load dataset
df = pd.read_csv('mental_health_processed.csv')

target_column = 'On a scale of 1–5, how would you rate your overall mental well-being right now?'

# Create new binary target column
# If the rating is 3 or less, it's 0 (Not Good). If it's more than 3 (i.e, 4 or 5), it's 1 (Good).
df['well_being_binary'] = np.where(df[target_column] <= 3, 0, 1)

# Set features (df_x) and NEW target (df_y)
df_x = df.drop(columns=[target_column, 'well_being_binary']) # Drop old and new target
df_y = df['well_being_binary'] # New target

# Split data
df_x_train, df_x_test, df_y_train, df_y_test = train_test_split(df_x, df_y, test_size=0.2, random_state=50)

# Initialize the Decision Tree model
dt = DecisionTreeClassifier(random_state=40)
dt.fit(df_x_train, df_y_train)

# Predict and get the accuracy
pred_dt = dt.predict(df_x_test)
accuracy = accuracy_score(df_y_test, pred_dt) * 100

print("Decision Tree Accuracy:")
print(accuracy)
```
